## 1. Klassen & Objekte – der Grundgedanke

Eine **Klasse** ist eine Schablone (Bauplan), ein **Objekt** ist eine konkrete Instanz davon.

Beispiel-Idee:
- Klasse: `MediaSession`
- Objekte: konkrete Nutzungs-Sessions (z. B. „Alex hört 30 Minuten Musik“)

In Python definierst du eine Klasse mit `class`:

````python
class MediaSession:
    def __init__(self, user, medium, minutes):
        self.user = user
        self.medium = medium
        self.minutes = minutes

    def describe(self):
        return f"{self.user} nutzt {self.medium} für {self.minutes} Minuten."

s1 = MediaSession("Alex", "Musik", 30)
s2 = MediaSession("Sam", "Film", 90)

print(s1.describe())
print(s2.describe())
````

Wichtige Punkte:

- `__init__` wird beim Erzeugen eines Objekts automatisch aufgerufen.
- `self` verweist auf **das aktuelle Objekt**.
- `self.user`, `self.medium`, `self.minutes` sind **Attribute** des Objekts.
- Methoden wie `describe` nutzen diese Attribute.

### Mini-Übung 1 – Eigene einfache Klasse

Definiere eine Klasse `UserProfile` mit:

- Attributen: `name`, `favorite_medium`, `minutes_per_week`
- einer Methode `summary()`, die einen String zurückgibt, z. B.:

`"Alex mag Musik und nutzt es 300 Minuten pro Woche."`

````python
class UserProfile:
    def __init__(self, name, favorite_medium, minutes_per_week):
        self.name = name
        self.favorite_medium = favorite_medium
        self.minutes_per_week = minutes_per_week

    def summary(self):
        return (
            f"{self.name} mag {self.favorite_medium} "
            f"und nutzt es {self.minutes_per_week} Minuten pro Woche."
        )

p = UserProfile("Alex", "Musik", 300)
print(p.summary())
````

## 2. Methoden – Verhalten in der Klasse

Klassen sind nicht nur Datencontainer – sie können auch **Verhalten** kapseln.

Beispiel: Wir erweitern `MediaSession` um eine einfache Methode, die Minuten in Stunden umrechnet.

````python
class MediaSession:
    def __init__(self, user, medium, minutes):
        self.user = user
        self.medium = medium
        self.minutes = minutes

    def describe(self):
        return f"{self.user} nutzt {self.medium} für {self.minutes} Minuten."

    def hours(self):
        return self.minutes / 60

session = MediaSession("Riley", "Game", 180)
print(session.describe())
print("Das sind", session.hours(), "Stunden.")
````

### Mini-Übung 2 – Methode hinzufügen

Erweitere `UserProfile` um eine Methode `add_minutes(delta)`, die `minutes_per_week` erhöht.

Beispiel:

```python
p = UserProfile("Alex", "Musik", 300)
p.add_minutes(60)  # jetzt 360
```

````python
class UserProfile:
    def __init__(self, name, favorite_medium, minutes_per_week):
        self.name = name
        self.favorite_medium = favorite_medium
        self.minutes_per_week = minutes_per_week

    def summary(self):
        return (
            f"{self.name} mag {self.favorite_medium} "
            f"und nutzt es {self.minutes_per_week} Minuten pro Woche."
        )

    def add_minutes(self, delta):
        if delta < 0:
            print("Warnung: negative Minuten werden ignoriert.")
            return
        self.minutes_per_week += delta

p = UserProfile("Alex", "Musik", 300)
print(p.summary())
p.add_minutes(60)
print(p.summary())
````

## 3. Listen von Objekten

Wie bei Dictionaries kannst du auch mit Listen von Objekten arbeiten.

Beispiel: mehrere `MediaSession`-Objekte in einer Liste sammeln und auswerten.

````python
sessions = [
    MediaSession("Alex", "Musik", 30),
    MediaSession("Alex", "Film", 60),
    MediaSession("Sam", "Game", 45),
]

for s in sessions:
    print(s.describe())

# Gesamtminuten pro Nutzer
total_per_user = {}
for s in sessions:
    total_per_user[s.user] = total_per_user.get(s.user, 0) + s.minutes

print("\nGesamtminuten pro Nutzer:")
for user, minutes in total_per_user.items():
    print(f"- {user}: {minutes} Minuten")
````

### Mini-Übung 3 – Profil-Liste

Lege drei `UserProfile`-Objekte an und speichere sie in einer Liste `profiles`.

Gib für jedes Profil die `summary()` aus und berechne die Summe aller `minutes_per_week`.

````python
profiles = [
    UserProfile("Alex", "Musik", 300),
    UserProfile("Sam", "Film", 200),
    UserProfile("Riley", "Game", 150),
]

total_minutes = 0
for p in profiles:
    print(p.summary())
    total_minutes += p.minutes_per_week

print("\nGesamtminuten aller Profile:", total_minutes)
````

## 4. Vererbung – spezialisierte Klassen

Mit **Vererbung** kannst du gemeinsame Funktionalität in einer Basisklasse bündeln
und spezialisierte Unterklassen davon ableiten.

Beispiel:
- Basisklasse `Medium`
- Unterklassen `Music`, `Video`, `Game`

Alle Medien haben z. B. einen `title`, aber weitere Eigenschaften unterscheiden sich.

````python
class Medium:
    def __init__(self, title):
        self.title = title

    def info(self):
        return f"Medium: {self.title}"

class Music(Medium):
    def __init__(self, title, artist):
        super().__init__(title)
        self.artist = artist

    def info(self):
        return f"Musik: '{self.title}' von {self.artist}"

class Video(Medium):
    def __init__(self, title, length_minutes):
        super().__init__(title)
        self.length_minutes = length_minutes

    def info(self):
        return f"Video: '{self.title}' ({self.length_minutes} Minuten)"

items = [
    Music("Song A", "Artist 1"),
    Video("Film B", 120),
]

for item in items:
    print(item.info())
````

### Mini-Übung 4 – Weitere Medienart

Füge eine weitere Unterklasse `Game` hinzu mit:

- Attributen: `title`, `platform`
- `info()`-Methode, z. B.:

`"Game: 'XYZ' auf PC"`

````python
class Game(Medium):
    def __init__(self, title, platform):
        super().__init__(title)
        self.platform = platform

    def info(self):
        return f"Game: '{self.title}' auf {self.platform}"

items.append(Game("Cooles Game", "PC"))

for item in items:
    print(item.info())
````

## 5. OOP und Datenstrukturen kombinieren

Du kannst OOP mit Listen/Dictionaries verbinden, um etwas strukturiertere Modelle zu bauen.

Beispiel: eine Klasse `MediaTracker`, die mehrere `MediaSession`-Objekte verwaltet.

````python
class MediaTracker:
    def __init__(self):
        self.sessions = []

    def add_session(self, session: MediaSession):
        self.sessions.append(session)

    def total_minutes(self):
        return sum(s.minutes for s in self.sessions)

    def total_minutes_per_user(self):
        totals = {}
        for s in self.sessions:
            totals[s.user] = totals.get(s.user, 0) + s.minutes
        return totals

tracker = MediaTracker()
tracker.add_session(MediaSession("Alex", "Musik", 30))
tracker.add_session(MediaSession("Alex", "Film", 60))
tracker.add_session(MediaSession("Sam", "Game", 45))

print("Gesamtminuten aller Sessions:", tracker.total_minutes())
print("Minuten pro Nutzer:", tracker.total_minutes_per_user())
````

## 6. Mini-Projekt – Einfaches Medienprofil-System

Baue ein kleines OOP-System mit folgenden Ideen (du kannst es vereinfachen oder erweitern):

- `UserProfile` kennt Name, Lieblingsmedium, Wochenminuten.
- `MediaSession` speichert einzelne Sessions.
- `MediaTracker` verwaltet alle Sessions.

Aufgabenidee:

1. Lege 2–3 `UserProfile`-Objekte an.
2. Füge mit `MediaTracker` zugehörige Sessions hinzu.
3. Berechne über den Tracker:
   - Gesamtminuten pro Nutzer
   - Liste aller Sessions eines Nutzers
4. Gib für jeden Nutzer einen kleinen Report aus, z. B.:

`"Alex: 3 Sessions, 180 Minuten insgesamt."`

````python
class MediaTrackerExtended(MediaTracker):
    def sessions_for_user(self, user):
        return [s for s in self.sessions if s.user == user]

profiles = [
    UserProfile("Alex", "Musik", 300),
    UserProfile("Sam", "Film", 200),
]

tracker2 = MediaTrackerExtended()
tracker2.add_session(MediaSession("Alex", "Musik", 60))
tracker2.add_session(MediaSession("Alex", "Game", 30))
tracker2.add_session(MediaSession("Sam", "Film", 120))

totals = tracker2.total_minutes_per_user()

for p in profiles:
    user_sessions = tracker2.sessions_for_user(p.name)
    count = len(user_sessions)
    total = totals.get(p.name, 0)
    print(f"{p.name}: {count} Sessions, {total} Minuten insgesamt.")
````
