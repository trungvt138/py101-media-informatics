## 1. Typische Fehlertypen in PythonEin paar häufige Fehlerarten:- `SyntaxError` – der Code ist syntaktisch ungültig (z. B. Klammer vergessen)- `NameError` – eine Variable ist nicht definiert- `TypeError` – Operation mit falschem Typ (z. B. Zahl + String)- `ValueError` – Wert hat den falschen Inhalt (z. B. `int("abc")`)- `IndexError` – Listenzugriff außerhalb des gültigen Bereichs- `KeyError` – Dictionary-Schlüssel existiert nichtViele dieser Fehler wirst du im Alltag ständig sehen. Das ist normal!

## 2. Tracebacks lesenWenn ein Fehler auftritt, zeigt Python einen **Traceback** – eine Rückverfolgung, wo der Fehler passiert ist.Beispiel (vereinfacht):```textTraceback (most recent call last):  File "main.py", line 3, in <module>    print(x)NameError: name 'x' is not defined```Wichtig:- Die **letzte Zeile** sagt dir den Fehlertyp und eine Beschreibung.- Die Zeilen darüber zeigen dir, **wo** im Code der Fehler entstanden ist.Lies Tracebacks immer **von unten nach oben**.

## 3. Fehler abfangen: `try` / `except`Statt dein Programm beim ersten Fehler abstürzen zu lassen, kannst du manche Fehler **abfangen** und eine eigene Reaktion definieren.Grundmuster:```pythontry:    # Code, der schiefgehen kann    value = int(input("Zahl: "))except ValueError:    print("Bitte eine gültige ganze Zahl eingeben!")```- Alles im `try`-Block wird „probiert“- Tritt z. B. ein `ValueError` auf, springt das Programm in den passenden `except`-Block- Danach läuft das Programm **weiter**, statt komplett abzubrechen

````python
# Beispiel: Sicher eine Zahl einlesentry:    text = input("Bitte eine ganze Zahl eingeben: ")    value = int(text)    print("Du hast die Zahl", value, "eingegeben.")except ValueError:    print("Das war keine gültige ganze Zahl.")
````

## 4. Mehrere Fehlertypen abfangenManchmal möchtest du mehrere Fehlertypen unterschiedlich behandeln.```pythontry:    # irgendein Code    result = data["key"] * 2except KeyError:    print("Schlüssel fehlt in data.")except TypeError:    print("Falscher Datentyp.")```Du kannst auch einen **allgemeinen** `except Exception:` nutzen, solltest dann aber aufpassen, nicht alles stumm zu schlucken.

````python
data = {"minutes": "30"}  # absichtlich '30' als Stringtry:    result = data["minutes"] * 2    print("Ergebnis:", result)except KeyError:    print("Schlüssel 'minutes' fehlt.")except TypeError:    print("Typfehler – prüfe, ob 'minutes' wirklich eine Zahl ist.")
````

## 5. Eigene Fehler auslösen: `raise`Manchmal willst du **bewusst** einen Fehler auslösen, wenn eine Bedingung verletzt ist.```pythondef set_volume(level):    if level < 0 or level > 100:        raise ValueError("Volume must be between 0 and 100")    # ...```Du kannst auch eigene Fehlertypen definieren (eigene Klassen), das brauchst du aber eher in größeren Projekten.

````python
def set_volume(level):    if level < 0 or level > 100:        raise ValueError("Volume must be between 0 and 100.")    print(f"Lautstärke gesetzt auf {level}%.")# Korrekte Nutzung:set_volume(30)# Falsche Nutzung (kommentiere ein, um den Fehler zu sehen):# set_volume(150)
````

## 6. Debugging-TaktikenFehlerbehandlung ist das eine – **Fehlersuche (Debugging)** das andere.Ein paar praktische Taktiken:1. **Fehlermeldung komplett lesen** – nicht nur die letzte Zeile überfliegen.2. **Print-Debugging**: Zwischenergebnisse mit `print(...)` ausgeben:   ```python   print("DEBUG: variable x =", x)   ```3. Code in kleinere Funktionen aufteilen und jede einzeln testen.4. In VS Code kannst du Breakpoints setzen und im Debugger schrittweise ausführen (F5 / Run & Debug).5. Eingabe-Daten prüfen: Stimmen Typen und Werte wirklich mit deiner Erwartung überein?Gerade am Anfang ist **Print-Debugging** ein völlig legitimer Weg.

````python
# Beispiel für Print-Debuggingdef engagement_score(minutes_music, minutes_video):    print("DEBUG: minutes_music =", minutes_music)    print("DEBUG: minutes_video =", minutes_video)    total = minutes_music + minutes_video    print("DEBUG: total =", total)    return min(100, total / 10)score = engagement_score(120, 45)print("Engagement-Score:", score)
````
