### Mini-Übung 3 – Lautstärke-AnzeigeSchreibe eine Funktion `print_volume(level, max_level=10)`, die eine einfache Lautstärke-Anzeige ausgibt, z. B.:```text[#####-----] 5/10```Hinweise:- Anzahl `#` entspricht `level`- Anzahl `-` entspricht `max_level - level`- Nutze z. B. String-Multiplikation: `"#" * 5` ergibt `"#####"`.

````python
# TODO: Mini-Übung 3 – print_volumedef print_volume(level, max_level=10):    if level < 0:        level = 0    if level > max_level:        level = max_level    hashes = "#" * level    dashes = "-" * (max_level - level)    bar = f"[{hashes}{dashes}] {level}/{max_level}"    print(bar)print_volume(5)print_volume(8, max_level=12)print_volume(level=3)
````

## 5. Scope: lokale vs. globale VariablenVariablen, die **in einer Funktion** definiert werden, sind **lokal**. Von außen kannst du sie nicht direkt verwenden.```pythondef foo():    x = 42  # lokal    print("In foo:", x)foo()print(x)  # Fehler!```Variablen außerhalb von Funktionen nennt man **global**. Sie sind im gesamten Modul sichtbar (mit Vorsicht zu verwenden).

````python
value = 10  # globale Variabledef multiply_by_two(n):    result = n * 2  # lokal    return resultprint("value:", value)print("value * 2:", multiply_by_two(value))# Die folgende Zeile würde einen Fehler auslösen, weil result lokal ist:# print(result)
````
